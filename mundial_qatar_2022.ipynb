```python
import pandas as pd


df = pd.read_csv("https://raw.githubusercontent.com/IgnacioPardo/Tecnologias_Exponenciales_2025/refs/heads/main/qatar.csv")


def fase_general(fase):
    if "Group" in fase:
        return "Fase de grupos"
    elif "Round of 16" in fase:
        return "Octavos"
    elif "Quarter" in fase:
        return "Cuartos"
    elif "Semi" in fase:
        return "Semifinal"
    else:
        return "Final"


df["fase_general"] = df["group"].apply(fase_general)


print("Cantidad de fases:", df["fase_general"].nunique())


print(df["fase_general"].unique())
```

```python
import pandas as pd


df = pd.read_csv("https://raw.githubusercontent.com/IgnacioPardo/Tecnologias_Exponenciales_2025/refs/heads/main/qatar.csv")


penales = {}


for _, fila in df.iterrows():

    equipo1 = fila["1"]
    equipo2 = fila["2"]

    penales[equipo1] = penales.get(equipo1, 0) + fila["1_panelties_scored"]
    penales[equipo2] = penales.get(equipo2, 0) + fila["2_panelties_scored"]


max_equipo = max(penales, key=penales.get)

print(max_equipo)
```

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/IgnacioPardo/Tecnologias_Exponenciales_2025/refs/heads/main/qatar.csv")
```

```python
print(df.head())
```

```python
equipo : str = "ARGENTINA"

filtroLocal : pd.Series = df["1"] == equipo

print(df[filtroLocal])
```

```python
equipo : str = "ARGENTINA"

filtroVisitante : pd.Series = df["2"] == equipo

print(df[filtroVisitante])
```

```python
partidos : pd.DataFrame = df[(df["1"] == equipo) | (df["2"] == equipo)]

print(partidos)
```

```python
goles : int = 0

for _, fila in partidos.iterrows():

    if fila["1"] == equipo:
        goles += fila["1_goals"]

    if fila["2"] == equipo:
        goles += fila["2_goals"]

print(goles)
```

```python
print(goles / len(partidos))
```

```python
ganados : int = 0

for _, fila in partidos.iterrows():

    if fila["1"] == equipo and fila["1_goals"] > fila["2_goals"]:
        ganados += 1

    elif fila["2"] == equipo and fila["2_goals"] > fila["1_goals"]:
        ganados += 1

print(ganados)
```

```python
equipos = {}

for _, fila in df.iterrows():

    equipos[fila["1"]] = equipos.get(fila["1"], 0) + fila["1_goals"]
    equipos[fila["2"]] = equipos.get(fila["2"], 0) + fila["2_goals"]

print(equipos)
```

```python
import matplotlib.pyplot as plt
```

```python
goles = list(equipos.values())

fig, ax = plt.subplots()

ax.bar(equipos.keys(), goles)

plt.xticks(rotation=90)

plt.show()
```

```python
for _, fila in df.iterrows():

    diferencia = abs(fila["1_goals"] - fila["2_goals"])

    print(fila["1"], "-", fila["2"], ":", diferencia)
```

```python
for _, fila in df.iterrows():

    if fila["1_goals"] > fila["2_goals"]:
        ganador = fila["1"]

    elif fila["2_goals"] > fila["1_goals"]:
        ganador = fila["2"]

    else:
        ganador = "Empate"

    print(fila["1"], "vs", fila["2"], "->", ganador)
```

```python
for fase in df["group"].unique():

    golesFase = {}

    partidosFase = df[df["group"] == fase]

    for _, fila in partidosFase.iterrows():

        golesFase[fila["1"]] = golesFase.get(fila["1"], 0) + fila["1_goals"]
        golesFase[fila["2"]] = golesFase.get(fila["2"], 0) + fila["2_goals"]

    mejor = max(golesFase, key=golesFase.get)

    print(fase, "->", mejor)
```
